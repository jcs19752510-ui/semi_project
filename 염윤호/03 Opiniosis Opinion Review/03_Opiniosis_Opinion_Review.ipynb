```python
import pandas as pd
import glob, os
import warnings
warnings.filterwarnings('ignore')
pd.set_option('display.max_colwidth', 700)


# 아래는 제 컴퓨터에서 압축 파일을 풀어 놓은 디렉터리이니, 여러분의 디렉터리를 설정해 주세요  
path = r'../data/topics'
# path로 지정한 디렉터리 밑에 있는 모든 .data 파일들의 파일명을 리스트로 취합
all_files = glob.glob(os.path.join(path, "*.data"))    
filename_list = []
opinion_text = []


# 개별 파일들의 파일명은 filename_list 리스트로 취합,
# 개별 파일들의 파일 내용은 DataFrame 로딩 후 다시 string으로 변환하여 opinion_text 리스트로 취합
for file_ in all_files:
    # 개별 파일을 읽어서 DataFrame으로 생성
    df = pd.read_table(file_,index_col=None, header=0,encoding='latin1')
   
    # 절대경로로 주어진 file 명을 가공. 만일 Linux에서 수행시에는 아래 \\를 / 변경.
    # 맨 마지막 .data 확장자도 제거
    filename_ = file_.split('\\')[-1]
    filename = filename_.split('.')[0]


    # 파일명 리스트와 파일 내용 리스트에 파일명과 파일 내용을 추가.
    filename_list.append(filename)
    opinion_text.append(df.to_string())


# 파일명 리스트와 파일 내용 리스트를  DataFrame으로 생성
document_df = pd.DataFrame({'filename':filename_list, 'opinion_text':opinion_text})
document_df.head()
```

```python
from nltk.stem import WordNetLemmatizer
import nltk
import string


remove_punct_dict = dict((ord(punct), None) for punct in string.punctuation)
lemmar = WordNetLemmatizer()


# 입력으로 들어온 token단어들에 대해서 lemmatization 어근 변환.
def LemTokens(tokens):
    return [lemmar.lemmatize(token) for token in tokens]


# TfidfVectorizer 객체 생성 시 tokenizer인자로 해당 함수를 설정하여 lemmatization 적용
# 입력으로 문장을 받아서 stop words 제거-> 소문자 변환 -> 단어 토큰화 -> lemmatization 어근 변환.
def LemNormalize(text):
    return LemTokens(nltk.word_tokenize(text.lower().translate(remove_punct_dict)))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer


tfidf_vect = TfidfVectorizer(tokenizer=LemNormalize, stop_words='english' , \
                             ngram_range=(1,2), min_df=0.05, max_df=0.85 )


#opinion_text 컬럼값으로 feature vectorization 수행
feature_vect = tfidf_vect.fit_transform(document_df['opinion_text'])
```

```python
from sklearn.cluster import KMeans


# 5개 집합으로 군집화 수행. 예제를 위해 동일한 클러스터링 결과 도출용 random_state=0
km_cluster = KMeans(n_clusters=5, max_iter=10000, random_state=0)
km_cluster.fit(feature_vect)
cluster_label = km_cluster.labels_
cluster_centers = km_cluster.cluster_centers_
```

```python
# 군집별 핵심 단어 추출(중심점에서 가장 가까운 단어가 핵심단어)
cluster_centers = km_cluster.cluster_centers_
print('cluster_centers shape :',cluster_centers.shape)  # (군집 수, 단어 개수) 축이 4610개, 중심점이 3개
print(cluster_centers)
```

```python
# 군집별 top n 핵심단어, 그 단어의 중심 위치 상대값, 대상 파일명들을 반환함.
def get_cluster_details(cluster_model, cluster_data, feature_names, clusters_num, top_n_features=10):
    cluster_details = {}
   
    # cluster_centers array 의 값이 큰 순으로 정렬된 index 값을 반환
    # 군집 중심점(centroid)별 할당된 word 피처들의 거리값이 큰 순으로 값을 구하기 위함.  
    centroid_feature_ordered_ind = cluster_model.cluster_centers_.argsort()[:,::-1]
   
    #개별 군집별로 iteration하면서 핵심단어, 그 단어의 중심 위치 상대값, 대상 파일명 입력
    for cluster_num in range(clusters_num):
        # 개별 군집별 정보를 담을 데이터 초기화.
        cluster_details[cluster_num] = {}
        cluster_details[cluster_num]['cluster'] = cluster_num
       
        # cluster_centers_.argsort()[:,::-1] 로 구한 index 를 이용하여 top n 피처 단어를 구함.
        top_feature_indexes = centroid_feature_ordered_ind[cluster_num, :top_n_features]
        top_features = [ feature_names[ind] for ind in top_feature_indexes ]
       
        # top_feature_indexes를 이용해 해당 피처 단어의 중심 위치 상댓값 구함
        top_feature_values = cluster_model.cluster_centers_[cluster_num, top_feature_indexes].tolist()
       
        # cluster_details 딕셔너리 객체에 개별 군집별 핵심 단어와 중심위치 상대값, 그리고 해당 파일명 입력
        cluster_details[cluster_num]['top_features'] = top_features
        cluster_details[cluster_num]['top_features_value'] = top_feature_values
        filenames = cluster_data[cluster_data['cluster_label'] == cluster_num]['filename']
        filenames = filenames.values.tolist()
        cluster_details[cluster_num]['filenames'] = filenames
       
    return cluster_details
```

```python
def print_cluster_details(cluster_details):
    for cluster_num, cluster_detail in cluster_details.items():
        print('####### Cluster {0}'.format(cluster_num))
        print('Top features:', cluster_detail['top_features'])
        print('Reviews 파일명 :',cluster_detail['filenames'][:7])
        print('==================================================')
```

### DBSCAN eps 자동 탐색

문서 수가 51개로 적어 `min_samples=3`으로 고정하고, 코사인 거리 기준 eps 후보들에 대해 군집 수 · 노이즈 비율 · 실루엣 점수를 계산한 뒤 실루엣 점수가 가장 높은 eps를 선택한다.

```python
from sklearn.cluster import DBSCAN
from sklearn.metrics import silhouette_score, pairwise_distances
import numpy as np

def eps_candidates(X):
    # 문서 쌍 코사인 거리의 5~55 백분위수를 eps 후보로 사용
    dist = pairwise_distances(X, metric='cosine')
    upper = dist[np.triu_indices_from(dist, k=1)]
    return sorted(set(np.percentile(upper, np.arange(5, 60, 5))))

def search_dbscan_eps(X, eps_grid, min_samples=3):
    rows = []
    best_eps, best_sil = None, -1
    for eps in eps_grid:
        labels = DBSCAN(eps=eps, min_samples=min_samples, metric='cosine').fit_predict(X)
        n_clusters = len(set(labels)) - (1 if -1 in labels else 0)
        noise_ratio = np.mean(labels == -1)
        sil = np.nan
        if n_clusters >= 2:
            mask = labels != -1
            if mask.sum() >= 2 and len(set(labels[mask])) >= 2:
                sil = silhouette_score(X[mask], labels[mask])
                if sil > best_sil:
                    best_eps, best_sil = eps, sil
        rows.append({'eps': eps, 'n_clusters': n_clusters, 'noise_ratio': noise_ratio, 'silhouette': sil})
    return best_eps, pd.DataFrame(rows)

eps_grid_tfidf = eps_candidates(feature_vect)
eps_grid_count = eps_candidates(count_feature_vect)

best_eps_tfidf, eps_search_tfidf = search_dbscan_eps(feature_vect, eps_grid_tfidf)
best_eps_count, eps_search_count = search_dbscan_eps(count_feature_vect, eps_grid_count)

print('TF-IDF 선택 eps :', round(best_eps_tfidf, 4))
print('Count 선택 eps :', round(best_eps_count, 4))
```

```python
import matplotlib.pyplot as plt

# 한글 폰트 설정 (Windows 기본 맑은 고딕) - 라벨/범례 깨짐 방지
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

# 시각화 색상: dataviz 가이드의 검증된 팔레트(카테고리 1번 blue) 사용
SERIES_BLUE = '#2a78d6'
REF_RED = '#e34948'

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
for ax, name, res, best_eps in zip(
    axes, ['TF-IDF', 'Count'], [eps_search_tfidf, eps_search_count], [best_eps_tfidf, best_eps_count]
):
    ax.plot(res['eps'], res['silhouette'], marker='o', color=SERIES_BLUE, label='실루엣 점수')
    ax.axvline(best_eps, color=REF_RED, linestyle='--', label=f'선택 eps={best_eps:.3f}')
    ax.set_title(f'{name} DBSCAN eps 탐색')
    ax.set_xlabel('eps (cosine distance)')
    ax.set_ylabel('silhouette score')
    ax.legend()
plt.tight_layout()
plt.show()
```

```python
# 선택된 eps로 최종 DBSCAN 수행
tfidf_dbscan_model = DBSCAN(eps=best_eps_tfidf, min_samples=3, metric='cosine')
document_df['label_tfidf_dbscan'] = tfidf_dbscan_model.fit_predict(feature_vect)

count_dbscan_model = DBSCAN(eps=best_eps_count, min_samples=3, metric='cosine')
document_df['label_count_dbscan'] = count_dbscan_model.fit_predict(count_feature_vect)

document_df[['filename', 'label_tfidf_kmeans', 'label_count_kmeans',
             'label_tfidf_dbscan', 'label_count_dbscan']].head()
```

### 4가지 조합 결과 비교

```python
combos = [
    ('TF-IDF + KMeans', feature_vect, document_df['label_tfidf_kmeans']),
    ('Count + KMeans', count_feature_vect, document_df['label_count_kmeans']),
    ('TF-IDF + DBSCAN', feature_vect, document_df['label_tfidf_dbscan']),
    ('Count + DBSCAN', count_feature_vect, document_df['label_count_dbscan']),
]

rows = []
for name, X, labels in combos:
    labels = labels.values
    n_clusters = len(set(labels)) - (1 if -1 in labels else 0)
    noise_ratio = np.mean(labels == -1)
    mask = labels != -1
    sil = silhouette_score(X[mask], labels[mask]) if len(set(labels[mask])) >= 2 else np.nan
    rows.append({'조합': name, '군집 수': n_clusters, '노이즈 비율': round(noise_ratio, 3),
                 '실루엣 점수': round(sil, 4) if pd.notna(sil) else np.nan})

comparison_df = pd.DataFrame(rows)
comparison_df
```

```python
# 카테고리 색상: dataviz 가이드의 검증된 팔레트(1~4번 슬롯)
CATEGORY_COLORS = ['#2a78d6', '#eb6834', '#1baf7a', '#eda100']

fig, ax = plt.subplots(figsize=(7, 4))
ax.bar(comparison_df['조합'], comparison_df['실루엣 점수'], color=CATEGORY_COLORS)
ax.set_ylabel('실루엣 점수')
ax.set_title('벡터화 × 군집화 조합별 실루엣 점수 비교')
ax.grid(axis='y', color='#e1e0d9', linewidth=0.8, zorder=0)
ax.set_axisbelow(True)
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

```python
from sklearn.decomposition import TruncatedSVD

# 카테고리 색상: dataviz 가이드의 검증된 팔레트(고정 순서, 8개 슬롯)
CLUSTER_COLORS = ['#2a78d6', '#eb6834', '#1baf7a', '#eda100',
                  '#e87ba4', '#008300', '#4a3aa7', '#e34948']

fig, axes = plt.subplots(2, 2, figsize=(11, 9))
for ax, (name, X, labels) in zip(axes.flat, combos):
    coords = TruncatedSVD(n_components=2, random_state=0).fit_transform(X)
    labels = labels.values
    for i, lab in enumerate(sorted(set(labels))):
        mask = labels == lab
        color = '#c3c2b7' if lab == -1 else CLUSTER_COLORS[i % len(CLUSTER_COLORS)]
        label_name = 'noise' if lab == -1 else f'cluster {lab}'
        ax.scatter(coords[mask, 0], coords[mask, 1], c=color, label=label_name, s=40)
    ax.set_title(name)
    ax.set_xlabel('SVD 1')
    ax.set_ylabel('SVD 2')
    ax.legend(fontsize=7)
plt.tight_layout()
plt.show()
```

## 군집 결과 기반 감성 분석 (VADER)

문서 단위(주제별 리뷰 취합본)로 감성 점수를 내면 문장 수가 많아 극단적으로 한쪽으로 쏠리므로, **원본 문장 단위**로 VADER 감성 점수를 계산한 뒤 문서(파일)별 긍정/중립/부정 비율로 집계한다.

```python
import nltk
nltk.download('vader_lexicon', quiet=True)
from nltk.sentiment import SentimentIntensityAnalyzer

sia = SentimentIntensityAnalyzer()

# 문장 단위 감성 분석을 위해 원본 파일을 다시 읽어 문장 리스트로 취합 (all_files 재사용)
sentence_records = []
for file_ in all_files:
    sent_df = pd.read_table(file_, index_col=None, header=0, encoding='latin1')
    filename = file_.split('\\')[-1].split('.')[0]
    for sentence in [sent_df.columns[0]] + sent_df.iloc[:, 0].astype(str).tolist():
        sentence_records.append({'filename': filename, 'sentence': sentence})

sentence_df = pd.DataFrame(sentence_records)
sentence_df['compound'] = sentence_df['sentence'].apply(lambda x: sia.polarity_scores(x)['compound'])

def classify_sentiment(compound):
    if compound >= 0.05:
        return 'positive'
    elif compound <= -0.05:
        return 'negative'
    return 'neutral'

sentence_df['sentiment_label'] = sentence_df['compound'].apply(classify_sentiment)
sentence_df['sentiment_label'].value_counts()
```

```python
# 문서(파일)별 긍정/중립/부정 비율을 집계하여 document_df에 병합
sentiment_ratio = (sentence_df.groupby('filename')['sentiment_label']
                    .value_counts(normalize=True)
                    .unstack(fill_value=0)
                    .rename(columns={'positive': 'positive_ratio',
                                     'neutral': 'neutral_ratio',
                                     'negative': 'negative_ratio'}))

document_df = document_df.merge(sentiment_ratio, left_on='filename', right_index=True)
document_df[['filename', 'positive_ratio', 'neutral_ratio', 'negative_ratio']].head()
```

### 군집별 감성 분포 (긍정/중립/부정)

```python
# 감성(긍정-중립-부정)은 양극 개념이므로 dataviz 가이드의 diverging 색상(blue<->red, 중립 gray)을 사용
SENTIMENT_COLORS = {'positive_ratio': '#2a78d6', 'neutral_ratio': '#f0efec', 'negative_ratio': '#e34948'}

fig, axes = plt.subplots(2, 2, figsize=(12, 9))
for ax, (name, _, labels) in zip(axes.flat, combos):
    tmp = document_df.copy()
    tmp['cluster'] = labels.values
    cluster_sentiment = tmp.groupby('cluster')[list(SENTIMENT_COLORS.keys())].mean()
    cluster_sentiment.index = [f'noise' if c == -1 else f'cluster {c}' for c in cluster_sentiment.index]
    cluster_sentiment.plot(kind='bar', stacked=True, ax=ax,
                            color=list(SENTIMENT_COLORS.values()), edgecolor='#fcfcfb', linewidth=1)
    ax.set_title(name)
    ax.set_ylabel('평균 감성 비율')
    ax.set_xlabel('')
    ax.legend(['positive', 'neutral', 'negative'], fontsize=7)
    ax.tick_params(axis='x', rotation=0)
plt.tight_layout()
plt.show()
```
