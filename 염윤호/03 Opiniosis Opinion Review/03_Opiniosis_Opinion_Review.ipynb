```python
import pandas as pd
import glob, os
import warnings
warnings.filterwarnings('ignore')
pd.set_option('display.max_colwidth', 700)


# 아래는 제 컴퓨터에서 압축 파일을 풀어 놓은 디렉터리이니, 여러분의 디렉터리를 설정해 주세요  
path = r'../data/topics'
# path로 지정한 디렉터리 밑에 있는 모든 .data 파일들의 파일명을 리스트로 취합
all_files = glob.glob(os.path.join(path, "*.data"))    
filename_list = []
opinion_text = []


# 개별 파일들의 파일명은 filename_list 리스트로 취합,
# 개별 파일들의 파일 내용은 DataFrame 로딩 후 다시 string으로 변환하여 opinion_text 리스트로 취합
for file_ in all_files:
    # 개별 파일을 읽어서 DataFrame으로 생성
    df = pd.read_table(file_,index_col=None, header=0,encoding='latin1')
   
    # 절대경로로 주어진 file 명을 가공. 만일 Linux에서 수행시에는 아래 \\를 / 변경.
    # 맨 마지막 .data 확장자도 제거
    filename_ = file_.split('\\')[-1]
    filename = filename_.split('.')[0]


    # 파일명 리스트와 파일 내용 리스트에 파일명과 파일 내용을 추가.
    filename_list.append(filename)
    opinion_text.append(df.to_string())


# 파일명 리스트와 파일 내용 리스트를  DataFrame으로 생성
document_df = pd.DataFrame({'filename':filename_list, 'opinion_text':opinion_text})
document_df.head()
```

```python
from nltk.stem import WordNetLemmatizer
import nltk
import string


remove_punct_dict = dict((ord(punct), None) for punct in string.punctuation)
lemmar = WordNetLemmatizer()


# 입력으로 들어온 token단어들에 대해서 lemmatization 어근 변환.
def LemTokens(tokens):
    return [lemmar.lemmatize(token) for token in tokens]


# TfidfVectorizer 객체 생성 시 tokenizer인자로 해당 함수를 설정하여 lemmatization 적용
# 입력으로 문장을 받아서 stop words 제거-> 소문자 변환 -> 단어 토큰화 -> lemmatization 어근 변환.
def LemNormalize(text):
    return LemTokens(nltk.word_tokenize(text.lower().translate(remove_punct_dict)))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer


tfidf_vect = TfidfVectorizer(tokenizer=LemNormalize, stop_words='english' , \
                             ngram_range=(1,2), min_df=0.05, max_df=0.85 )


#opinion_text 컬럼값으로 feature vectorization 수행
feature_vect = tfidf_vect.fit_transform(document_df['opinion_text'])
```

```python
# 군집별 핵심 단어 추출(중심점에서 가장 가까운 단어가 핵심단어)
cluster_centers = km_cluster.cluster_centers_
print('cluster_centers shape :',cluster_centers.shape)  # (군집 수, 단어 개수) 축이 4610개, 중심점이 3개
print(cluster_centers)
```

```python
# 군집별 top n 핵심단어, 그 단어의 중심 위치 상대값, 대상 파일명들을 반환함.
def get_cluster_details(cluster_model, cluster_data, feature_names, clusters_num, top_n_features=10):
    cluster_details = {}
   
    # cluster_centers array 의 값이 큰 순으로 정렬된 index 값을 반환
    # 군집 중심점(centroid)별 할당된 word 피처들의 거리값이 큰 순으로 값을 구하기 위함.  
    centroid_feature_ordered_ind = cluster_model.cluster_centers_.argsort()[:,::-1]
   
    #개별 군집별로 iteration하면서 핵심단어, 그 단어의 중심 위치 상대값, 대상 파일명 입력
    for cluster_num in range(clusters_num):
        # 개별 군집별 정보를 담을 데이터 초기화.
        cluster_details[cluster_num] = {}
        cluster_details[cluster_num]['cluster'] = cluster_num
       
        # cluster_centers_.argsort()[:,::-1] 로 구한 index 를 이용하여 top n 피처 단어를 구함.
        top_feature_indexes = centroid_feature_ordered_ind[cluster_num, :top_n_features]
        top_features = [ feature_names[ind] for ind in top_feature_indexes ]
       
        # top_feature_indexes를 이용해 해당 피처 단어의 중심 위치 상댓값 구함
        top_feature_values = cluster_model.cluster_centers_[cluster_num, top_feature_indexes].tolist()
       
        # cluster_details 딕셔너리 객체에 개별 군집별 핵심 단어와 중심위치 상대값, 그리고 해당 파일명 입력
        cluster_details[cluster_num]['top_features'] = top_features
        cluster_details[cluster_num]['top_features_value'] = top_feature_values
        filenames = cluster_data[cluster_data['cluster_label'] == cluster_num]['filename']
        filenames = filenames.values.tolist()
        cluster_details[cluster_num]['filenames'] = filenames
       
    return cluster_details
```

```python
def print_cluster_details(cluster_details):
    for cluster_num, cluster_detail in cluster_details.items():
        print('####### Cluster {0}'.format(cluster_num))
        print('Top features:', cluster_detail['top_features'])
        print('Reviews 파일명 :',cluster_detail['filenames'][:7])
        print('==================================================')
```

### DBSCAN eps 자동 탐색

문서 수가 51개로 적어 `min_samples=3`으로 고정하고, 코사인 거리 기준 eps 후보들에 대해 군집 수 · 노이즈 비율 · 실루엣 점수를 계산한 뒤 실루엣 점수가 가장 높은 eps를 선택한다.

```python
from sklearn.cluster import DBSCAN
from sklearn.metrics import silhouette_score, pairwise_distances
import numpy as np

def eps_candidates(X):
    # 문서 쌍 코사인 거리의 5~55 백분위수를 eps 후보로 사용
    dist = pairwise_distances(X, metric='cosine')
    upper = dist[np.triu_indices_from(dist, k=1)]
    return sorted(set(np.percentile(upper, np.arange(5, 60, 5))))

def search_dbscan_eps(X, eps_grid, min_samples=3):
    rows = []
    best_eps, best_sil = None, -1
    for eps in eps_grid:
        labels = DBSCAN(eps=eps, min_samples=min_samples, metric='cosine').fit_predict(X)
        n_clusters = len(set(labels)) - (1 if -1 in labels else 0)
        noise_ratio = np.mean(labels == -1)
        sil = np.nan
        if n_clusters >= 2:
            mask = labels != -1
            if mask.sum() >= 2 and len(set(labels[mask])) >= 2:
                sil = silhouette_score(X[mask], labels[mask])
                if sil > best_sil:
                    best_eps, best_sil = eps, sil
        rows.append({'eps': eps, 'n_clusters': n_clusters, 'noise_ratio': noise_ratio, 'silhouette': sil})
    return best_eps, pd.DataFrame(rows)

eps_grid_tfidf = eps_candidates(feature_vect)
eps_grid_count = eps_candidates(count_feature_vect)

best_eps_tfidf, eps_search_tfidf = search_dbscan_eps(feature_vect, eps_grid_tfidf)
best_eps_count, eps_search_count = search_dbscan_eps(count_feature_vect, eps_grid_count)

print('TF-IDF 선택 eps :', round(best_eps_tfidf, 4))
print('Count 선택 eps :', round(best_eps_count, 4))
```

### 5가지 조합 결과 비교

```python
combos = [
    ('TF-IDF + KMeans', feature_vect, document_df['label_tfidf_kmeans']),
    ('Count + KMeans', count_feature_vect, document_df['label_count_kmeans']),
    ('Count(정규화) + KMeans', count_feature_vect_norm, document_df['label_count_norm_kmeans']),
    ('TF-IDF + DBSCAN', feature_vect, document_df['label_tfidf_dbscan']),
    ('Count + DBSCAN', count_feature_vect, document_df['label_count_dbscan']),
]

rows = []
for name, X, labels in combos:
    labels = labels.values
    n_clusters = len(set(labels)) - (1 if -1 in labels else 0)
    noise_ratio = np.mean(labels == -1)
    mask = labels != -1
    sil = silhouette_score(X[mask], labels[mask]) if len(set(labels[mask])) >= 2 else np.nan
    rows.append({'조합': name, '군집 수': n_clusters, '노이즈 비율': round(noise_ratio, 3),
                 '실루엣 점수': round(sil, 4) if pd.notna(sil) else np.nan})

comparison_df = pd.DataFrame(rows)
comparison_df
```

## 군집 결과 기반 감성 분석 (VADER)

문서 단위(주제별 리뷰 취합본)로 감성 점수를 내면 문장 수가 많아 극단적으로 한쪽으로 쏠리므로, **원본 문장 단위**로 VADER 감성 점수를 계산한 뒤 문서(파일)별 긍정/중립/부정 비율로 집계한다.

```python
import nltk
nltk.download('vader_lexicon', quiet=True)
from nltk.sentiment import SentimentIntensityAnalyzer

sia = SentimentIntensityAnalyzer()

# 문장 단위 감성 분석을 위해 원본 파일을 다시 읽어 문장 리스트로 취합 (all_files 재사용)
sentence_records = []
for file_ in all_files:
    sent_df = pd.read_table(file_, index_col=None, header=0, encoding='latin1')
    filename = file_.split('\\')[-1].split('.')[0]
    for sentence in [sent_df.columns[0]] + sent_df.iloc[:, 0].astype(str).tolist():
        sentence_records.append({'filename': filename, 'sentence': sentence})

sentence_df = pd.DataFrame(sentence_records)
sentence_df['compound'] = sentence_df['sentence'].apply(lambda x: sia.polarity_scores(x)['compound'])

def classify_sentiment(compound):
    if compound >= 0.05:
        return 'positive'
    elif compound <= -0.05:
        return 'negative'
    return 'neutral'

sentence_df['sentiment_label'] = sentence_df['compound'].apply(classify_sentiment)
sentence_df['sentiment_label'].value_counts()
```

```python
# 문서(파일)별 긍정/중립/부정 비율을 집계하여 document_df에 병합
sentiment_ratio = (sentence_df.groupby('filename')['sentiment_label']
                    .value_counts(normalize=True)
                    .unstack(fill_value=0)
                    .rename(columns={'positive': 'positive_ratio',
                                     'neutral': 'neutral_ratio',
                                     'negative': 'negative_ratio'}))

document_df = document_df.merge(sentiment_ratio, left_on='filename', right_index=True)
document_df[['filename', 'positive_ratio', 'neutral_ratio', 'negative_ratio']].head()
```

## 차원 축소(TruncatedSVD) 효과 검증 (dimensionality_reduction_comparison.md 적용)

TF-IDF 벡터에 TruncatedSVD를 적용하는 것이 K-means 군집 품질을 실제로 개선하는지 검증한다.

- **파이프라인 A (baseline)**: TF-IDF → KMeans
- **파이프라인 B (SVD)**: TF-IDF → TruncatedSVD → Normalizer → KMeans

⚠️ 서로 다른 차원의 공간에서 계산한 실루엣 점수는 직접 비교하지 않는다(고차원일수록 거리 집중 현상으로 실루엣이 구조적으로 낮게 나와 SVD가 불공정하게 유리해짐). 대신 **대리 정답 기반 외부 평가(ARI/NMI)** 를 주력으로, **군집 안정성**과 **군집 키워드 육안 확인**을 보조 지표로 사용한다.

### 선행 작업 — 파일명 형식 확인 후 대리 정답(proxy label) 정의

```python
# 파일명 형식을 먼저 직접 확인한다
sample_names = sorted(os.path.basename(f) for f in all_files)
print(len(sample_names), '개 파일')
sample_names[:10]
```

```python
# 파일명 = {주제(aspect)}_{대상(product)} 구조. 대상 식별자를 접미사로 매칭해 aspect/product/category를 분리한다.
# 원본 데이터의 표기 불일치 보정: netbook_1005ha/asus_netbook_1005ha, swissotel_chicago/swissotel_hotel_chicago 는
# 각각 동일 대상이므로 같은 이름으로 정규화한다.
PRODUCT_MAP = {
    'asus_netbook_1005ha': ('netbook_1005ha', 'electronics'),
    'netbook_1005ha': ('netbook_1005ha', 'electronics'),
    'garmin_nuvi_255W_gps': ('garmin_nuvi_255W_gps', 'electronics'),
    'amazon_kindle': ('amazon_kindle', 'electronics'),
    'ipod_nano_8gb': ('ipod_nano_8gb', 'electronics'),
    'windows7': ('windows7', 'electronics'),
    'honda_accord_2008': ('honda_accord_2008', 'car'),
    'toyota_camry_2007': ('toyota_camry_2007', 'car'),
    'swissotel_hotel_chicago': ('swissotel_chicago', 'hotel'),
    'swissotel_chicago': ('swissotel_chicago', 'hotel'),
    'bestwestern_hotel_sfo': ('bestwestern_hotel_sfo', 'hotel'),
    'holiday_inn_london': ('holiday_inn_london', 'hotel'),
}
ASPECT_ALIAS = {'room': 'rooms'}  # room/rooms 표기 불일치 보정
_suffixes = sorted(PRODUCT_MAP.keys(), key=len, reverse=True)  # 긴 접미사부터 매칭(예: asus_netbook_1005ha 우선)

def parse_proxy_labels(filename):
    for suf in _suffixes:
        if filename == suf or filename.endswith('_' + suf):
            product, category = PRODUCT_MAP[suf]
            aspect = filename[: -(len(suf) + 1)] if filename != suf else ''
            aspect = ASPECT_ALIAS.get(aspect, aspect)
            return pd.Series({'target_product': product, 'target_category': category, 'aspect_label': aspect})
    raise ValueError(f'매칭되는 대상을 찾지 못함: {filename}')

document_df[['target_product', 'target_category', 'aspect_label']] = document_df['filename'].apply(parse_proxy_labels)
print('대상 종류(target_category) 기준:')
print(document_df['target_category'].value_counts())
print()
print('주제(aspect) 기준 고유 개수:', document_df['aspect_label'].nunique())
```

### 4-5절 — 파이프라인 A/B 실행, 대리 정답 기반 외부 평가(ARI/NMI) + 군집 안정성

`n_clusters=3`(기존 메인 분석과 동일하게 고정), `random_state` 30개 시드로 반복하여 평균을 낸다. 같은 30회 반복 결과를 이용해 쌍별 ARI 평균(군집 안정성)도 함께 계산한다.

```python
from sklearn.decomposition import TruncatedSVD
from sklearn.preprocessing import Normalizer
from sklearn.pipeline import make_pipeline
from sklearn.metrics import adjusted_rand_score, normalized_mutual_info_score

N_CLUSTERS = 3
N_SEEDS = 30
N_COMPONENTS_LIST = [2, 5, 10, 20, 30, 40]

target_labels = document_df['target_category'].values
aspect_labels = document_df['aspect_label'].values

def run_pipeline_seeds(X, n_seeds=N_SEEDS, n_clusters=N_CLUSTERS):
    return [KMeans(n_clusters=n_clusters, max_iter=10000, random_state=seed, n_init=10).fit_predict(X)
            for seed in range(n_seeds)]

def summarize_runs(runs, true_target, true_aspect):
    ari_t = np.mean([adjusted_rand_score(true_target, r) for r in runs])
    nmi_t = np.mean([normalized_mutual_info_score(true_target, r) for r in runs])
    ari_a = np.mean([adjusted_rand_score(true_aspect, r) for r in runs])
    nmi_a = np.mean([normalized_mutual_info_score(true_aspect, r) for r in runs])
    pair_scores = [adjusted_rand_score(runs[i], runs[j])
                   for i in range(len(runs)) for j in range(i + 1, len(runs))]
    stability = np.mean(pair_scores)
    return ari_t, nmi_t, ari_a, nmi_a, stability

svd_results = []

# 파이프라인 A: TF-IDF -> KMeans (baseline)
runs_a = run_pipeline_seeds(feature_vect)
ari_t, nmi_t, ari_a, nmi_a, stability = summarize_runs(runs_a, target_labels, aspect_labels)
svd_results.append({'파이프라인': 'A (TF-IDF)', 'n_components': np.nan,
                     'ARI_대상종류': ari_t, 'NMI_대상종류': nmi_t,
                     'ARI_주제': ari_a, 'NMI_주제': nmi_a, '안정성': stability})

# 파이프라인 B: TF-IDF -> TruncatedSVD -> Normalizer -> KMeans
svd_pipelines = {}
for k in N_COMPONENTS_LIST:
    lsa = make_pipeline(TruncatedSVD(n_components=k, random_state=0), Normalizer(copy=False))
    X_svd = lsa.fit_transform(feature_vect)
    runs_b = run_pipeline_seeds(X_svd)
    svd_pipelines[k] = (lsa, X_svd, runs_b)
    ari_t, nmi_t, ari_a, nmi_a, stability = summarize_runs(runs_b, target_labels, aspect_labels)
    svd_results.append({'파이프라인': 'B (SVD)', 'n_components': k,
                         'ARI_대상종류': ari_t, 'NMI_대상종류': nmi_t,
                         'ARI_주제': ari_a, 'NMI_주제': nmi_a, '안정성': stability})

comparison_svd_df = pd.DataFrame(svd_results).round(4)
comparison_svd_df
```

```python
# n_components 상한 판단 근거: 문서 51개이므로 유의미한 축은 최대 50개 근처.
# 상위 40개 축의 누적 설명분산비를 확인한다.
svd_diag = TruncatedSVD(n_components=40, random_state=0).fit(feature_vect)
cum_var = np.cumsum(svd_diag.explained_variance_ratio_)
print('n_components=10 누적 설명분산비 :', round(cum_var[9], 3))
print('n_components=20 누적 설명분산비 :', round(cum_var[19], 3))
print('n_components=40 누적 설명분산비 :', round(cum_var[39], 3))
```

```python
# n_components에 따른 ARI 변화 시각화 (baseline은 기준선으로 표시)
svd_only = comparison_svd_df[comparison_svd_df['파이프라인'] == 'B (SVD)']
baseline_row = comparison_svd_df[comparison_svd_df['파이프라인'] == 'A (TF-IDF)'].iloc[0]

fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(svd_only['n_components'], svd_only['ARI_대상종류'], marker='o', color='#2a78d6', label='ARI (대상 종류)')
ax.plot(svd_only['n_components'], svd_only['ARI_주제'], marker='o', color='#eb6834', label='ARI (주제)')
ax.axhline(baseline_row['ARI_대상종류'], color='#2a78d6', linestyle='--', alpha=0.6, label='baseline ARI (대상 종류)')
ax.axhline(baseline_row['ARI_주제'], color='#eb6834', linestyle='--', alpha=0.6, label='baseline ARI (주제)')
ax.set_xlabel('n_components (TruncatedSVD)')
ax.set_ylabel('평균 ARI (시드 30회 평균)')
ax.set_title('TF-IDF vs SVD 파이프라인의 대리 정답 기반 외부 평가')
ax.legend(fontsize=8)
ax.grid(axis='y', color='#e1e0d9', linewidth=0.8, zorder=0)
ax.set_axisbelow(True)
plt.tight_layout()
plt.show()
```

### 6절 — 군집 키워드 육안 확인

ARI가 가장 높은 SVD n_components에 대해 군집 중심을 `inverse_transform`으로 단어 공간에 되돌려 핵심 단어를 확인하고, 기존에 계산해 둔 baseline(TF-IDF+KMeans) 군집 키워드와 비교한다.

```python
best_k = int(comparison_svd_df.loc[comparison_svd_df['n_components'].notna(), :]
                              .sort_values('ARI_대상종류', ascending=False).iloc[0]['n_components'])
print('ARI(대상 종류)가 가장 높은 n_components :', best_k)

lsa_best, X_svd_best, _ = svd_pipelines[best_k]
km_best = KMeans(n_clusters=N_CLUSTERS, max_iter=10000, random_state=0, n_init=10)
km_best.fit(X_svd_best)

# SVD 잠재 축은 사람이 읽을 수 없으므로 단어 공간으로 역투영
svd_step = lsa_best.named_steps['truncatedsvd']
centroids_word_space = svd_step.inverse_transform(km_best.cluster_centers_)

print(f'--- 파이프라인 B (SVD n_components={best_k}) 군집별 핵심 단어 ---')
for i, c in enumerate(centroids_word_space):
    top = c.argsort()[::-1][:10]
    print(f'Cluster {i}:', [feature_names[j] for j in top])

print()
print('--- 파이프라인 A (baseline TF-IDF) 군집별 핵심 단어 ---')
print_cluster_details(get_cluster_details(cluster_model=tfidf_kmeans_model, cluster_data=document_df,
                                           feature_names=feature_names, clusters_num=N_CLUSTERS, top_n_features=10))
```

**결론**

- **대상 종류(전자기기/자동차/호텔) 기준**: baseline(TF-IDF+KMeans)이 이미 ARI 0.997 수준으로 거의 완벽하게 분리한다. TF-IDF 벡터 자체가 도메인별로 어휘가 뚜렷이 갈리는 데이터라 개선의 여지가 크지 않다. 그중 `n_components=5`인 SVD가 ARI 1.000·안정성 1.000으로 baseline보다 소폭 더 낫고 완전히 안정적인 반면, `n_components=10~40`은 오히려 baseline보다 낮은 ARI를 보였다. 즉 **SVD는 성분 수를 작게(5 근처) 잡을 때만 미세한 개선 효과가 있고, 성분 수를 늘릴수록 이득이 사라지거나 역전된다.**
- **주제(aspect, 35개 세부 클래스) 기준**: baseline과 SVD 모두 ARI가 0.05 미만으로 낮다. 이는 `n_clusters=3`으로는 애초에 35개 세부 주제를 구분할 수 없기 때문이며, 두 파이프라인 간 우열을 가리는 근거로 쓸 수 없다.
- **군집 키워드 육안 확인**: baseline과 SVD(n_components=5) 모두 `battery/screen/keyboard`(전자기기), `mileage/seat/comfort`(자동차), `room/service/staff`(호텔) 로 군집 내용이 깨끗하게 갈려 성공 예에 해당한다.
- **종합 판단**: 이 데이터셋(51개 문서, 도메인이 뚜렷한 어휘)에서는 TruncatedSVD 도입이 필수가 아니다. baseline이 이미 충분히 좋고, SVD는 성분 수를 잘 고르면 아주 소폭의 안정성/정확도 이득을 줄 뿐 극적인 개선은 아니다. 문서 수가 많고 어휘 중첩이 큰 데이터셋일수록 SVD의 효과가 더 크게 나타날 가능성이 높다.
