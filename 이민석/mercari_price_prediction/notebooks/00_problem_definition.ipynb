# Mercari 상품 가격 분석 및 예측

## 01. 문제 정의

### 1.1 프로젝트 배경

온라인 중고거래 플랫폼에서는 상품의 특성과 시장 상황에 따라 적절한 판매 가격을 설정하는 것이 중요하다. 
가격이 지나치게 높을 경우 상품의 판매 가능성이 낮아질 수 있으며, 반대로 지나치게 낮은 가격은 상품의 잠재적인 가치를 충분히 반영하지 못할 수 있다.

본 프로젝트에서는 Mercari 상품 데이터를 활용하여 상품의 다양한 특성과 가격 간의 관계를 분석하고, 
상품 정보를 기반으로 가격을 예측하는 머신러닝 모델을 구축하고자 한다.

단순한 가격 예측에 그치지 않고 통계적 분석과 머신러닝을 함께 활용하여 
가격 결정 요인을 해석하고 예측 모델의 성능을 비교하는 것을 목표로 한다.


### 1.2 비즈니스 문제

Mercari 상품 등록 과정에서 상품의 특성을 바탕으로 적절한 가격 수준을 추정할 수 있다면 
판매자의 가격 설정을 지원하고 플랫폼의 상품 가격 분석에도 활용할 수 있다.

따라서 다음과 같은 비즈니스 질문을 설정하였다.

- 상품의 어떤 특성이 판매 가격에 영향을 미치는가?
- 상품의 특성을 이용하여 적정 가격을 예측할 수 있는가?
- 가격 예측에 가장 중요한 상품 정보는 무엇인가?
- 텍스트 정보가 가격 예측 성능 향상에 기여하는가?


### 1.3 통계적 질문

통계적 분석을 통해 상품의 특성과 가격 간의 관계를 검증하고자 한다.

- 상품 상태에 따라 가격에 유의한 차이가 존재하는가?
- 카테고리와 브랜드는 가격에 유의한 영향을 미치는가?
- 배송비 부담 여부에 따라 가격 수준에 차이가 존재하는가?
- 여러 변수를 동시에 고려했을 때 각 변수의 영향력은 어떻게 달라지는가?
- 다중회귀모형의 통계적 가정은 적절하게 충족되는가?

이를 위해 탐색적 데이터 분석(EDA) 이후 다중회귀분석을 수행하고, 
회귀계수, 유의확률, 신뢰구간, 결정계수 및 다중공선성 등을 검토한다.


### 1.4 머신러닝 질문

통계적 분석을 통해 확인한 변수 관계를 바탕으로 머신러닝 모델을 구축하여 
상품 가격의 예측 가능성을 평가한다.

- 상품의 특성만으로 가격을 어느 정도 정확하게 예측할 수 있는가?
- 선형 회귀모델과 비선형 머신러닝 모델의 예측 성능에는 차이가 있는가?
- Ridge와 Lasso와 같은 정규화 회귀모델은 일반 선형회귀보다 안정적인 예측 성능을 보이는가?
- 상품명과 상품 설명과 같은 텍스트 정보가 가격 예측 성능을 향상시키는가?
- 머신러닝 모델이 중요하게 판단하는 변수는 통계적 회귀분석에서 유의한 변수와 어떻게 다른가?


### 1.5 프로젝트 목표

본 프로젝트의 목표는 다음과 같다.

1. Mercari 상품 데이터의 구조와 품질을 파악한다.
2. 탐색적 데이터 분석을 통해 상품 특성과 가격의 관계를 파악한다.
3. 통계적 회귀분석을 통해 가격 결정 요인의 유의성을 검증한다.
4. 다양한 머신러닝 회귀모델을 구축하고 예측 성능을 비교한다.
5. 텍스트 정보를 활용하여 가격 예측 성능의 개선 가능성을 검토한다.
6. Feature Importance 및 SHAP 등의 방법을 활용하여 모델의 예측 근거를 해석한다.
7. 통계적 추론과 머신러닝 예측 결과를 종합하여 가격 결정 요인에 대한 인사이트를 도출한다.


### 1.6 핵심 분석 질문

본 프로젝트에서는 다음 세 가지 핵심 질문에 답하고자 한다.

> **Q1. 상품의 어떤 특성이 가격에 유의한 영향을 미치는가?**

> **Q2. 상품의 특성을 이용하여 가격을 얼마나 정확하게 예측할 수 있는가?**

> **Q3. 통계적으로 확인된 가격 결정 요인과 머신러닝 모델이 중요하게 판단한 변수는 어떻게 다른가?**
