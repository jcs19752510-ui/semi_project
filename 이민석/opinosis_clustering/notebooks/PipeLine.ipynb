# 4. 군집분석 및 감성 분석

## 4.1 분석 목적

본 분석에서는 제품 리뷰 문장에 포함된 **주요 군집(주제)을 추출하고, 각 군집이 감성에 미치는 영향을 분석**하였다.

구체적인 분석 목적은 다음과 같다.

1. 전체 리뷰에서 어떤 군집(주제)이 존재하는지 파악
2. 군집별 평균 감성 점수의 차이 분석
3. 회귀계수를 기준으로 감성에 가장 큰 영향을 미치는 군집 파악
4. 동일한 군집이라도 제품군에 따라 감성 차이가 존재하는지 확인

---

## 4.2 분석 파이프라인

전체 분석 과정은 다음과 같이 구성하였다.

```text
51개 리뷰 파일
      ↓
① 데이터 로드 및 기본 노이즈 정제
      ↓
② VADER 감성 분석
   └─ 원문 기준 문장별 감성 점수 계산
      ↓
③ 군집분석용 텍스트 전처리
   └─ 감성어 및 불용어 제거
      ↓
④ TF-IDF 벡터화
      ↓
⑤ TruncatedSVD 차원축소
      ↓
⑥ 엘보우 방법으로 최적 군집 수(K) 탐색
      ↓
⑦ KMeans 군집분석
      ↓
⑧ 군집별 주요 단어 확인 (군집 중심 기준)
      ↓
⑨ 사람이 군집 라벨링
      ↓
⑩ Ridge 회귀
   └─ 군집 → 감성 점수 관계 분석 (해석용)
      ↓
⑪ OLS 회귀 (statsmodels, HC3 강건표준오차)
   └─ 통계적 유의성 검정 (p-value)
      ↓
⑫ VIF 다중공선성 점검 및 단계적 제거
   └─ 통제변수 정제 후 재적합, 강건성 검토
      ↓
⑬ 위계적 회귀 (군집 블록 F검정)
   └─ 군집 요인 자체의 유의성 검증
      ↓
⑭ 군집 안정성 검토
   └─ 실루엣 점수, ARI 재현성
      ↓
⑮ 시각화 및 해석
   ├─ 군집별 평균 감성 점수
   ├─ 군집별 회귀계수
   ├─ 군집 × 제품 히트맵
   └─ 체르노프 페이스 (종합 시각화)
```

## 4.3 데이터 규모 및 감성 점수 확인

총 **7,086개 문장**을 대상으로 분석을 진행하였다. 데이터는 **13개 제품, 36개 속성(파일)**으로 구성되며, 51개 리뷰 파일이 정상적으로 로드되었다.

문장별 감성 점수는 VADER를 사용하여 원문을 기준으로 계산하였다.

VADER 감성 점수의 극단값을 확인한 결과, 실제 문장의 감성 방향과 일치하는 모습을 보였다.

* 가장 긍정적인 문장: **0.984**

  * "fun car to drive", "comfortable", "looking forward to a lot of great miles" 등 강한 긍정 표현 포함
* 가장 부정적인 문장: **-0.92**

  * "not fixable", "would not purchase another camry", "disappointed" 등 강한 부정 표현 포함

따라서 VADER의 감성 점수가 리뷰 문장의 전반적인 긍정·부정 방향을 비교적 적절하게 반영하고 있음을 확인하였다.

## 4.4 최적 군집 수(K) 결정

KMeans 군집분석을 위해 군집 수(K)를 **2~30** 범위에서 탐색하였다.

각 K에 대해 KMeans의 inertia(군집 내 분산)를 계산한 뒤, inertia 곡선에서 **첫 지점과 마지막 지점을 잇는 직선으로부터 가장 멀리 떨어진 지점**을 엘보우(꺾이는 지점)로 자동 판단하는 방식을 사용하였다.

그 결과 **K=12**가 최적 군집 수로 선택되었다.

이는 초기 분석에서 임의로 설정했던 K=12와 일치하지만, 이번에는 데이터 기반 탐색을 통해 결정된 값이라는 점에서 의미가 있다.

따라서 이후 분석에서는 **12개의 군집을 최종 군집 구조로 사용**하였다.

## 4.6 군집별 평균 감성

군집별 평균 감성 점수를 비교함으로써 특정 주제가 언급될 때 리뷰의 전반적인 감성 방향이 어떻게 나타나는지 확인하였다.

특히 호텔 관련 군집에서는 **직원 응대와 위치 관련 군집이 상대적으로 높은 긍정 감성**을 보였으며, 자동차 관련 군집 중에서는 **시트 및 변속기 관련 군집의 감성이 가장 낮게 나타났다.**

다만 군집은 특정 단어의 등장 패턴을 기준으로 구성되기 때문에, 평균 감성만으로 해당 주제가 직접적으로 긍정 또는 부정의 원인이라고 단정해서는 안 된다.

## 4.7 Ridge 회귀를 이용한 군집과 감성의 관계 분석

군집이 감성 점수에 미치는 방향과 상대적인 영향력을 확인하기 위해 **Ridge 회귀**를 적용하였다. 군집 라벨은 원-핫 인코딩하여 회귀의 독립변수로 사용하였다.

회귀모형의 설명력은 다음과 같다.

* **R² = 0.15**
* **RMSE = 0.41**

R²가 0.15라는 것은 군집 정보만으로 문장 감성 점수 변동의 약 **15%를 설명할 수 있음**을 의미한다.

예측 모델 관점에서는 높은 설명력이라고 보기 어렵지만, 본 분석의 목적은 높은 예측 정확도를 확보하는 것이 아니라 **어떤 군집이 감성과 어떤 방향으로 연결되는지 해석하는 것**에 있다.

동일한 군집에서도 표현 방식에 따라 감성이 달라질 수 있다. 예를 들어 배터리라는 동일한 주제를 언급하더라도 "배터리가 오래간다"는 긍정적인 표현이고, "배터리가 금방 닳는다"는 부정적인 표현이다.

따라서 R²가 높지 않은 것은 자연스러운 결과이며, **군집 외에도 개별 단어와 문맥, 표현 방식 등이 감성에 중요한 영향을 미친다는 점**을 보여준다.

## 4.8 감성을 높이는 군집

Ridge 회귀계수가 양수인 군집 중 영향력이 큰 군집은 다음과 같다.

| 순위 | 군집         |  회귀계수 | 해석            |
| -- | ---------- | ----: | ------------- |
| 1  | cluster_8  | +0.21 | 호텔 위치 + 직원 응대 |
| 2  | cluster_11 | +0.19 | 호텔 프런트/컨시어지   |
| 3  | cluster_3  | +0.16 | 배터리 지속시간      |

특히 **cluster_8과 cluster_11이 높은 양의 회귀계수**를 보였다.

이는 호텔 리뷰에서 **직원 응대, 프런트, 컨시어지 등 사람이 직접 제공하는 서비스와 관련된 내용이 긍정적인 감성과 연결되는 경향**이 있음을 의미한다.

또한 cluster_3의 경우 배터리 지속시간과 관련된 군집으로, 배터리 사용시간에 대한 긍정적인 평가가 감성 점수를 높이는 방향으로 나타난 것으로 해석할 수 있다.

## 4.9 감성을 낮추는 군집

반대로 음의 회귀계수를 갖는 군집 중 영향력이 큰 군집은 다음과 같다.

| 순위 | 군집         |  회귀계수 | 해석         |
| -- | ---------- | ----: | ---------- |
| 1  | cluster_1  | -0.41 | 자동차 시트/변속기 |
| 2  | cluster_0  | -0.08 | 혼합/애매한 주제  |
| 3  | cluster_10 | -0.07 | 자동차 연비     |

가장 눈에 띄는 군집은 **cluster_1**이다.

cluster_1의 회귀계수는 **-0.41**로 다른 군집에 비해 절대값이 크게 나타났다.

해당 군집의 주요 단어는 `seat`, `transmission`, `leather`, `driver seat`, `shift` 등이며, 자동차의 **좌석 편의성과 변속감**에 대한 내용이 중심이다.

실제 원문에서도 `uncomfortable`, `kills my back`, VCM 변속 이질감 등 불편함이나 불만을 나타내는 내용이 확인되었다.

따라서 **자동차 시트의 편안함과 변속 품질이 해당 제품군에서 중요한 부정적 평가 요인으로 나타났다**고 해석할 수 있다.

## 4.10 군집 해석 시 주의사항

### ① cluster_7과 cluster_8의 주제 중복

cluster_7과 cluster_8은 모두 `location`, `wharf` 등의 단어를 포함하고 있어 사실상 **호텔 위치와 관련된 유사한 군집**으로 볼 수 있다.

특히 `fisherman wharf`, `tube`, `gloucester`와 같은 제품별 고유명사가 포함되면서 하나의 위치 관련 주제가 여러 군집으로 분리된 것으로 판단된다.

따라서 해석 단계에서는 두 군집을 별개의 완전히 독립적인 주제로 보기보다는 **'호텔 위치'라는 상위 주제 안에서 세분화된 군집**으로 보는 것이 적절하다.

### ② cluster_0의 해석 제한

cluster_0은 `interior`, `battery`, `kindle`, `direction` 등 서로 다른 제품과 관련된 단어가 혼합되어 있다.

따라서 명확한 주제라고 보기 어려운 **혼합형 군집**에 해당한다.

이 때문에 cluster_0의 회귀계수인 **-0.08에 지나치게 큰 의미를 부여하는 것은 적절하지 않다.**

## 4.11 체르노프 페이스를 통한 종합 시각화

앞서 확인한 군집별 통계치(문장 수, 평균 감성, 감성 표준편차, 회귀계수)를 하나의 그림으로 종합하기 위해 **체르노프 페이스(Chernoff Face)**를 이용한 시각화를 추가로 수행하였다.

체르노프 페이스는 다변량 데이터를 사람의 얼굴 특징에 매핑하여 여러 지표를 동시에 직관적으로 비교할 수 있도록 하는 시각화 기법이다. 각 얼굴 요소는 다음과 같이 매핑하였다.

| 얼굴 요소 | 매핑 데이터 | 해석 |
| --- | --- | --- |
| 얼굴 크기 | 군집 내 문장 수 | 클수록 리뷰에서 많이 언급된 주제 |
| 눈 크기 | 감성 점수 표준편차 | 클수록 해당 주제에 대한 의견이 긍정·부정으로 혼재 |
| 눈썹 각도 | 회귀계수 | 위로 갈수록 감성을 강하게 끌어올리는 군집, 아래로 갈수록 강하게 끌어내리는 군집 |
| 입 모양 | 평균 감성 점수 | 웃는 정도가 클수록 대체로 긍정적으로 언급되는 주제 |
| 얼굴 색 | 회귀계수 부호 | 초록(긍정 영향) / 빨강(부정 영향) / 회색(중립) |

시각화 결과, **cluster_1(자동차 시트/변속기)**은 눈썹이 아래로 처지고 입이 찡그려진 붉은색 얼굴로 나타나 가장 강한 부정적 군집임을 한눈에 확인할 수 있었다. 반대로 **cluster_8(호텔 위치+응대)**은 눈썹이 위로 올라가고 활짝 웃는 초록색 얼굴로 나타나, 회귀계수·평균 감성 분석 결과와 일관된 패턴을 시각적으로도 확인하였다.

이처럼 체르노프 페이스는 여러 개의 수치 지표를 표나 개별 그래프로 나열하지 않고도, 군집 간 차이를 직관적으로 비교할 수 있게 해준다는 점에서 보조적인 해석 도구로 활용하였다.

## 4.12 통계적 유의성 검정 (OLS)

Ridge 회귀는 정규화(regularization)를 적용하는 방법이므로 p-value와 같은 통계적 유의성 지표를 제공하지 않는다. 따라서 군집이 감성에 미치는 영향이 통계적으로 유의한지 검증하기 위해, statsmodels를 이용한 **OLS(최소제곱법) 회귀**를 추가로 적합하였다.

각 범주형 변수(군집·제품·속성)는 첫 범주를 기준범주로 제거(drop_first)하여 완전 다중공선성(더미변수 함정)을 우선 해소하였으며, 잔차의 이분산성 가능성을 고려하여 **HC3 강건표준오차(robust standard error)**를 적용하였다.

회귀모형의 설명력은 다음과 같다.

* **R² = 0.143**
* **Adj. R² = 0.136**
* **F(58, 7025) = 27.08, p < .001**

Ridge 결과와 비교하면 계수의 방향과 상대적 크기는 유사하게 유지되었으나, OLS를 통해 비로소 **어떤 군집이 통계적으로 유의한지**를 명확히 확인할 수 있었다. 특히 cluster_5, 6, 4, 10은 계수 자체는 0이 아니었지만 p>.1로 **통계적으로 유의하지 않아**, Ridge 계수만으로 이들을 "감성에 영향을 미치는 군집"으로 해석하는 것은 신중해야 함을 보여준다.

## 4.13 다중공선성 점검 및 정제 (VIF)

OLS 계수의 신뢰도를 확보하기 위해 통제변수(제품·속성) 간 다중공선성을 VIF(분산팽창계수)로 점검하였다.

1차 점검 결과, 전체 58개 변수 중 **16개 변수가 VIF≥10(심각)**으로 나타났다. 이 중 일부는 완전 다중공선성(VIF=∞)을 보였는데, 원인을 추적한 결과 원본 데이터의 파일명 표기 불일치(예: 제품명이 다르게 기재된 파일)로 인해 특정 속성·제품 조합이 항상 함께 등장하는 구조적 결함으로 확인되었다.

다행히 **군집(cluster) 변수 자체의 VIF는 모두 1.6~6.5 범위**로 양호하여, 다중공선성 문제는 본 연구의 핵심 관심 변수가 아닌 통제변수 쪽에 국한되어 있었다.

이에 따라 군집 변수는 보존한 채, VIF≥10인 통제변수를 **단계적 제거(stepwise elimination)** 방식으로 하나씩 제거하며 재계산하였다. 그 결과 **4개 변수만 제거**되었고(전체의 6.9%), 정제 후에는 **VIF≥10인 변수가 0개**로 완전히 해소되었다.

정제 후 재적합한 OLS 모형의 군집 계수를 1차 결과와 비교하면 다음과 같다.

| 군집 | 1차 계수 (p) | 정제 후 계수 (p) | 변화폭 |
| --- | ---: | ---: | ---: |
| cluster_8 | 0.3115 (\*\*\*) | 0.3268 (\*\*\*) | +0.015 |
| cluster_11 | 0.3084 (\*\*\*) | 0.3179 (\*\*\*) | +0.010 |
| cluster_3 | 0.2337 (\*\*\*) | 0.2346 (\*\*\*) | +0.001 |
| cluster_9 | 0.1303 (\*\*\*) | 0.1385 (\*\*\*) | +0.008 |
| cluster_7 | 0.1223 (\*\*\*) | 0.1367 (\*\*\*) | +0.014 |
| cluster_2 | 0.1217 (.) | 0.1207 (.) | -0.001 |
| cluster_1 | -0.2970 (\*\*\*) | -0.2956 (\*\*\*) | +0.001 |

모든 군집에서 계수 변화가 0.001~0.015 수준으로 미미하였으며, 유의성 판정(부호 및 유의수준)이 뒤바뀐 군집은 없었다. 이는 **통제변수 간 다중공선성이 본 연구의 핵심 결과(군집이 감성에 미치는 영향)를 왜곡하지 않았음**을 정량적으로 보여주는 강건성 검토(robustness check) 결과로 해석할 수 있다.

## 4.14 위계적 회귀분석 (군집 블록 F검정)

개별 군집 계수의 유의성과는 별개로, **"군집이라는 분류 체계 자체가 통계적으로 유의한 설명력을 추가하는가"**를 검증하기 위해 위계적 회귀분석(hierarchical regression)을 수행하였다.

* **축소모형(restricted model)**: 통제변수(제품·속성)만 포함
* **완전모형(full model)**: 축소모형 + 군집 변수 11개(기준범주 제외)

두 모형의 설명력 차이를 부분 F검정으로 비교한 결과는 다음과 같다.

| 모형 | R² |
| --- | ---: |
| 축소모형 (통제변수만) | 0.1177 |
| 완전모형 (통제변수+군집) | 0.1416 |
| ΔR² | 0.0239 |

* **F(11, 7018) = 17.81, p < .001**

즉 제품·속성 정보만으로 설명되는 부분(11.77%)을 넘어서, 군집 정보를 추가함으로써 감성 점수 변동의 **2.39%p를 통계적으로 유의하게 추가 설명**하였다. 이는 본 연구에서 도출한 군집 구조가 제품·속성 정보로 환원되지 않는 **독자적인 설명력을 지닌 요인**임을 시사한다.

## 4.15 군집 안정성 검토 (실루엣 점수 및 재현성)

군집 구조 자체의 타당성을 점검하기 위해 두 가지 안정성 지표를 추가로 확인하였다.

**① 실루엣 점수(Silhouette Score)**: K=12 군집의 응집도와 분리도를 나타내는 지표로, -1~1 범위를 가지며 통상 0.25 미만은 군집 간 경계가 뚜렷하지 않음을 의미한다. 본 분석의 실루엣 점수는 **0.064**로 낮게 나타나, 군집 간 경계가 기하학적으로는 뚜렷하게 분리되지 않았음을 확인하였다.

**② 재현성(Adjusted Rand Index, ARI)**: 서로 다른 초기화(random_state)로 KMeans를 10회 반복하고, 최초 채택한 군집 결과와의 일치도를 ARI로 측정하였다(1에 가까울수록 두 결과가 일치).

| 지표 | 값 |
| --- | ---: |
| 평균 ARI | 0.586 |
| 표준편차 | 0.135 |
| 최소값 | 0.320 |
| 최대값 | 0.801 |

평균 ARI 0.586은 "중간 수준의 안정성"에 해당하며, 초기화에 따라 군집의 대체적인 구조는 유지되지만 일부 군집 경계는 유동적임을 의미한다. 이는 4.10에서 관찰한 **cluster_7과 cluster_8이 모두 '호텔 위치'라는 유사한 주제를 나누어 담고 있는 현상**과 일맥상통하는 결과로, 리뷰 문장이 짧고 여러 주제가 혼재되어 언급되는 텍스트 데이터의 특성상 나타나는 자연스러운 한계로 해석된다.

다만 주목할 점은, 이러한 기하학적 분리의 한계에도 불구하고 **4.14의 위계적 회귀분석에서 군집 변수 블록이 통계적으로 유의한 추가 설명력을 보였다는 것**이다. 이는 군집 경계가 완벽하지 않음에도 불구하고, 군집화 결과가 통계적으로 실질적인 정보를 담고 있음을 시사한다.

## 4.16 한계 및 향후 연구

본 분석의 통계적 보강 과정에서 확인된 한계점과 이에 대한 향후 연구 방향은 다음과 같다.

1. **군집 간 경계의 불완전성**: 실루엣 점수(0.064)와 ARI 재현성(평균 0.586)이 낮게 나타나, 현재의 hard clustering(KMeans) 방식이 리뷰 문장에 혼재된 다중 주제를 완전히 분리하지 못하는 한계가 있다. 향후 연구에서는 문장이 여러 주제에 동시에 속할 수 있음을 반영하는 **연성 군집화(soft clustering)**, 예컨대 LDA와 같은 토픽모델링 기법을 적용하여 이를 보완할 필요가 있다.
2. **VADER 감성사전의 도메인 특이성**: VADER는 영어 규칙 기반 범용 감성사전으로, 제품 리뷰 도메인 특유의 표현에 대한 정밀도가 검증되지 않았다. 향후 연구에서는 사람이 직접 라벨링한 표본과의 일치도(정확도, Cohen's Kappa 등)를 검증하는 절차가 필요하다.
3. **원본 데이터의 표기 불일치**: 51개 리뷰 파일 중 일부에서 제품명 표기가 통일되지 않아(예: 동일 제품이 서로 다른 명칭으로 기재됨) 통제변수 간 완전 다중공선성이 발생하였다. 본 연구에서는 VIF 기반 단계적 제거로 대응하였으나, 근본적으로는 원본 데이터 정제 단계에서 표기를 표준화하는 것이 바람직하다.
4. **횡단면적 분석의 한계**: 본 분석은 특정 시점에 수집된 리뷰 문장을 대상으로 하였으며, 시간에 따른 감성 변화나 제품 개선 효과 등은 반영하지 못하였다.

## 4.17 종합 해석

이번 군집분석을 통해 리뷰 데이터에서 **자동차, 전자제품, 호텔 서비스 및 시설**과 관련된 다양한 주제가 존재함을 확인하였다.

Ridge 및 OLS 회귀분석 결과, 군집별 감성 영향에는 뚜렷한 차이가 나타났다.

* **호텔 직원 응대 및 프런트 관련 군집**은 긍정적인 감성과 강하게 연결되었으며 통계적으로도 유의하였다(p<.001)
* **배터리 지속시간** 역시 긍정적인 방향으로 유의하게 연결되었다
* **자동차 시트 및 변속기 관련 군집**은 가장 강한 부정적 영향을 보였으며 통계적으로 유의하였다(p<.001)
* **GPS 속도제한 표시, 호텔 객실, 화면 크기, 연비 관련 군집**은 계수 자체는 존재하였으나 통계적으로 유의하지 않아, 해석에 신중을 기해야 한다

다중공선성 점검(VIF) 결과 통제변수 간 일부 문제가 발견되었으나, 단계적 제거 후 재적합한 결과 군집 계수는 안정적으로 유지되어 본 연구의 핵심 결과가 다중공선성으로부터 강건함을 확인하였다. 또한 위계적 회귀분석을 통해 군집이라는 요인 자체가 제품·속성 정보로 환원되지 않는 통계적으로 유의한 설명력을 지님을 검증하였다(F=17.81, p<.001).

다만 군집만으로 감성의 약 14% 정도를 설명할 수 있었으며, 군집 안정성 검토 결과 군집 간 경계가 완전히 뚜렷하지는 않았다는 점에서, **군집 자체가 감성을 결정한다기보다 특정 주제가 어떤 맥락과 표현으로 언급되는지가 감성에 더 큰 영향을 미친다**고 볼 수 있다.

종합하면, 본 연구는 (1) 데이터 기반으로 결정된 군집 수, (2) Ridge와 OLS를 병행한 해석 및 통계적 검정, (3) VIF 기반 다중공선성 정제와 강건성 검토, (4) 위계적 회귀를 통한 군집 요인의 유의성 검증, (5) 실루엣·ARI를 통한 군집 안정성 점검을 모두 거쳐, 단순 탐색적 분석을 넘어 통계적으로 뒷받침되는 결론에 도달하였다. 다만 군집 경계의 불완전성 등 한계점 역시 명확히 존재하므로, 결과는 **대표 단어, 실제 리뷰 문장, 제품군별 분포, 체르노프 페이스와 같은 종합 시각화, 그리고 본 절의 통계적 검정 결과를 함께 고려하여 해석하는 것이 적절하다.**
